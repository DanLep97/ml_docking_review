```python
import pandas
from plotly.subplots import make_subplots
import plotly.graph_objects as go
from qbiolip import load_db
from datetime import datetime
import random
import re
df = load_db("../../databases/QBioLiP/qbiolip.csv", "../../databases/ec_uniprot_ids.txt", build_cols=False)
df = df[df.target]

# calculate number of atoms in ligands:
def count_res(x):
    return len(x.split("-"))
df["num_atoms"] = df.lig_atoms.str.len()

# Create subplots
fig = make_subplots(
    rows=1, 
    cols=2,
    subplot_titles=[
        "Number of atoms in ligands", 
        "Number of residues in biological assemblies"
    ]
)

# Add first histogram with custom number of bins
fig.add_trace(
    go.Histogram(x=df["num_atoms"], name="Atoms in ligands", nbinsx=30),  # Change 30 to your desired number of bins
    row=1, col=1
)

# Add second histogram with filtered data
fig.add_trace(
    go.Histogram(x=df["rec_len_resi"], name="Residues in biological assemblies"),
    row=1, col=2
)

# Set log scale for second y-axis
# fig.update_yaxes(type="log", row=1, col=2)
fig.update_xaxes(title_text="Number of atoms", row=1, col=1)
fig.update_yaxes(title_text="Count", row=1, col=1)
fig.update_xaxes(title_text="Number of residues", row=1, col=2)

# Update layout
fig.update_layout(
    font_size=21,
    height=500,
    width=800,
    showlegend=False,
    margin=dict(t=0,b=0,l=0,r=0), 
)

fig.show()
fig.write_image("../results/figures/data_distribution.jpg", height=500, width=800)
```
