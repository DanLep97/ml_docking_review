```python
import sys
import torch
sys.path.append("../preprocess")
import plotly.express as px
import plotly.graph_objects as go
from qbiolip import load_db
from pprint import pprint
import pandas
from copy import copy
from plotly.subplots import make_subplots
from matplotlib.gridspec import GridSpec

df = load_db("../../databases/QBioLiP/qbiolip.csv", "../../databases/ec_uniprot_ids.txt")
df = df.loc[df.target, [
    "ligand_detail",
    "pdb_id",
    "assembly_id",
    "ligand_id",
    "group"
]]
outputs = torch.load("../../databases/results/posebusters_all_df.pt", weights_only=False)
result_dfs = outputs["bust_df"]
tools_ordered = [
    "Boltz-2", "AlphaFold3", "DynamicBind", 
    "DiffDock", "TankBind", "NeuralPLexer", "EquiBind", "DPL", "Vina"]
result_dfs = {tool: result_dfs[tool] for tool in tools_ordered}
cols = [
    "ligand_detail",
    "blind_docking",
    "bond_lengths",
    "bond_angles",
    "aromatic_ring_maximum_distance_from_plane",
    "double_bond_maximum_distance_from_plane",
    "internal_steric_clash",
    "minimum_distance_to_protein",
    "rmsd",
    "centroid_distance",
    "volume_overlap_protein"
]

targets = [
    "rmsd",
    "centroid_distance",
    "bond_lengths",
    "bond_angles",
    "aromatic_ring_maximum_distance_from_plane",
    "double_bond_maximum_distance_from_plane",
    "internal_steric_clash",
    "minimum_distance_to_protein",
]
final_df_data = {}
for c in targets:
    final_df_data[c] = []
    final_df_data[f"{c}_std"] = []

tool_names = []

# cols for final df
blind_dockings = []
enz_classes = []
tools = []
counts = []
failed_names = []
merged_dfs = []
for i, (tool, full_results) in enumerate(result_dfs.items()):
    assert all(full_results.tool == tool), f"not all results come from {tool}"
    tool_names.append(tool)
    results = full_results.loc[
        (full_results.mol_pred_loaded) &
        (full_results.mol_true_loaded) &
        (full_results.mol_cond_loaded) &  # generally, all types of molecules are loaded
        (~full_results["rmsd_≤_2å"].isna()) &
        (~full_results.rmsd.isna()) &
        (~full_results.centroid_distance.isna())&
        (full_results.sanitization), 
        cols
    ]
    print(f"For tool {tool}, n = {results.index.get_level_values(0).to_series(name='name').unique().shape[0]}")
    # sanity check:
    results_names_not_in_df = results.loc[~results.ligand_detail.isin(df.ligand_detail.tolist())].ligand_detail.values
    if len(results_names_not_in_df) > 0:
        print("Len of cases:", len(results_names_not_in_df))
        print("Targets from DB:", df.ligand_detail.values[:10])
        print("Targets from posebusters:", results.ligand_detail.values[:10])
        print("Targets not in both:", results_names_not_in_df[:10])
    # merge results with original df, based on ligand_detail:
    results = results.merge(df, on="ligand_detail", how="inner")
    merged_dfs.append({"tool":tool, "df":results})
    # groupby:
    groups = results.groupby(["blind_docking", "group"])
    for j, ((blind_docking, enz_class), state) in enumerate(groups):
        for col in targets:
            rows = state[col]
            if rows.dtype != "bool":
                metric = rows.mean()
                std = rows.std()
            else: # if it's a boolean, get the fraction of positives
                metric = torch.tensor(rows.values, dtype=torch.float).mean().item()
                std = 0
            final_df_data[col].append(metric)
            final_df_data[f"{col}_std"].append(std)
        blind_dockings.append(blind_docking)
        enz_classes.append(enz_class)
        tools.append(tool)
        counts.append(state.shape[0])
final_df_data["enz_class"] = enz_classes
final_df_data["blind_docking"] = blind_dockings
final_df_data["tool"] = tools
final_df_data["state_count"] = counts
final_df = pandas.DataFrame(final_df_data)
final_df.fillna(0, inplace=True)

# organize in tensors in order to get the max value
all_arr = []
all_std = []
class_counts = []
for i, tool in enumerate(result_dfs.keys()):
    arr = []
    std = []
    for j, col in enumerate(targets):
        all_rows = final_df.loc[final_df.tool == tool]
        for k in [True,False]:
            docking_rows = all_rows.loc[all_rows.blind_docking == k]
            for l in ["A", "B", "C"]:
                rows = docking_rows.loc[docking_rows.enz_class == l]
                # metric info:
                arr.append(rows[col].values[0])
                std.append(rows[f"{col}_std"].values[0])
                # count:
                if j == 0:
                    class_counts.append(rows.state_count.item())
    all_arr.append(arr)
    all_std.append(std)
all_arr = torch.tensor(all_arr)

maxs = all_arr.max(dim=0)
max_ind = maxs.indices
max_vals = maxs.values

mins = all_arr.min(dim=0)
min_ind = mins.indices
min_vals = mins.values
table_str = ""

# SUPPLEMENTARY INFORMATION TABLES:
# write the latex syntax for RMSD
table_str += "RMSD table \n"
for i in range(all_arr.shape[0]):
    table_str += f"{tool_names[i]}"
    # add values:
    for j, n in enumerate(all_arr[i][:6]):
        table_str += f"&\\textbf{{{n:.1f}}}" if i == min_ind[j] else f"&{n:.1f}" # add values
        table_str += f"$\pm$ {all_std[i][j]:.1f}"
    table_str += "\\\\ \n"
table_str += "\n\n"

# write the latex syntax for Centroid
table_str += "Centroid table \n"
for i in range(all_arr.shape[0]):
    table_str += f"{tool_names[i]}"
    # add values:
    for j, n in enumerate(all_arr[i][6:12], 6):
        table_str += f"&\\textbf{{{n:.1f}}}" if i == min_ind[j] else f"&{n:.1f}" # add values
        table_str += f" $\pm$ {all_std[i][j]:.1f}"
    table_str += "\\\\ \n"
table_str += "\n\n"

table_str += "Count table \n"
# write the latex syntax for Counts
for i in range(all_arr.shape[0]):
    table_str += f"{tool_names[i]}"
    for idx in class_counts[i*6:i*6+6]:
        table_str += f"&{idx}"# add counts
    table_str += "\\\\ \n"
table_str += "\n\n"

# write the latex syntax for Angles, Bonds
table_str += "Angles, bonds table \n"
for i in range(all_arr.shape[0]):
    table_str += f"{tool_names[i]}"
    for j, n in enumerate(all_arr[i]):
        ind = max_ind
        if j >= 12 and j < 24:
            table_str += f"&\\textbf{{{n:.2f}}}" if i == ind[j] or n == 1 else f"&{n:.2f}"
    table_str += "\\\\ \n"
table_str += "\n\n"

table_str += "Rings, Double Bonds \n"
# write the latex syntax for Rings, Double Bonds, Clashes
for i in range(all_arr.shape[0]):
    table_str += f"{tool_names[i]}"
    for j, n in enumerate(all_arr[i]):
        if j >= 24 and j < 36:
            if j < 24:
                ind = max_ind
                vals = max_vals
            if j >= 24:
                ind = min_ind
                vals = min_vals
            table_str += f"&\\textbf{{{n:.2f}}}" if i == ind[j] or n == vals[j] else f"&{n:.2f}"
    table_str += "\\\\ \n"
table_str += "\n\n"

# write the latex syntax for Min dist to protein
table_str += "Clashes, Min dist to protein\n"
for i in range(all_arr.shape[0]):
    table_str += f"{tool_names[i]}"
    for j, n in enumerate(all_arr[i]):
        ind = max_ind
        if j >= 36:
            table_str += f"&\\textbf{{{n:.2f}}}" if i == ind[j] or n == 1 else f"&{n:.2f}"
    table_str += "\\\\ \n"
table_str += "\n\n"

with open("table.txt", "w") as table_f:
    table_f.write(table_str)
```

```python
from rdkit import Chem
from rdkit.Chem import AllChem
from rdkit.Chem import DataStructs
import pickle

threshold_tools = ["DynamicBind", "DiffDock", "DPL", "TankBind", "EquiBind"]
fg_gen = AllChem.GetRDKitFPGenerator()

QBioLiP_ligand_pdb_paths = glob.glob("../../databases/PDBbindv2020/pdb/*/*_ligand.sdf")
QBioLiP_ligand_fp = []
for ligand_path in QBioLiP_ligand_pdb_paths:
    mol = Chem.SDMolSupplier(ligand_path)[0]
    if mol:
        fp = fg_gen.GetFingerprint(mol)
        QBioLiP_ligand_fp.append(fp)

DB_ligand_pdb_paths = glob.glob("../../databases/QBioLiP/nr_lig/*.pdb")
DB_ligand_pdb_paths = [p for p in DB_ligand_pdb_paths if "randconf" not in p]
DB_ligand_details = pandas.Series([
    p.split("/")[-1].replace(".pdb","") for p in DB_ligand_pdb_paths
])
ccd_ligand_fp = {}
for i,ligand_path in enumerate(DB_ligand_pdb_paths):
    mol = Chem.MolFromPDBFile(ligand_path)
    ccd = DB_ligand_details[i].split("_")[2]
    if mol:
        fp = fg_gen.GetFingerprint(mol)
        ccd_ligand_fp[ccd] = fp

ccd_tanimoto = {ccd:torch.tensor([
        DataStructs.TanimotoSimilarity(ccd_fp, QBioLiP_fp)
        for QBioLiP_fp in QBioLiP_ligand_fp
    ]) for ccd,ccd_fp in ccd_ligand_fp.items()
}
```

```python
# cluster RMSD by threshold values:
import matplotlib.pyplot as plt

thresholds = [30,60,90,99,100]
seq_df = pandas.read_csv("similarity_full.tsv", sep="\t")

all_seq_avg = []
all_seq_counts = []
all_substrate_avg = []
all_substrate_counts = []

for i, tool in enumerate(threshold_tools):

    DB_df = [df["df"] for df in merged_dfs if df["tool"] == tool][0]

    seq_avg = {"PBD":[], "TBD":[]}
    seq_counts = {"PBD":[], "TBD":[]}
    substrate_avg = {"PBD":[], "TBD":[]}
    substrate_counts = {"PBD":[], "TBD":[]}

    for threshold in thresholds:
        # do enzymes:
        if threshold != 100:
            t_df = seq_df.loc[
                (seq_df.identity_pct <= threshold) &
                (seq_df.identity_pct > 0)
            ]
        else:
            t_df = seq_df.loc[(seq_df.identity_pct <= threshold)]
        u_pdb_ids = [p[:4] for p in t_df.test_id.unique()]
        for regime in ["PBD", "TBD"]:
            is_bd = regime == "TBD"
            threshold_df = DB_df.loc[
                (DB_df.pdb_id.isin(u_pdb_ids)) &
                (DB_df.blind_docking == is_bd)
            ] 
            hits = threshold_df.loc[threshold_df.rmsd <= 5]
            seq_avg[regime].append((hits.shape[0]/threshold_df.shape[0])*100)
            seq_counts[regime].append(threshold_df.shape[0])

        # do substrates:
        t = threshold/100
        tanimoto_matrix = torch.stack([
            ccd_tanimoto[ccd] for ccd in DB_df.ligand_id.values
        ])
        similarity_matrix_nonzero = (tanimoto_matrix <= t).nonzero()[:,0]
        threshold_output_idx = (
            similarity_matrix_nonzero.bincount() == tanimoto_matrix.shape[1]
        ).nonzero()[:,0].tolist()
        for regime in ["PBD", "TBD"]:
            is_bd = regime == "TBD"
            threshold_df = DB_df.loc[
                (DB_df.index.isin(threshold_output_idx)) &
                (DB_df.blind_docking == is_bd)
            ] 
            hits = threshold_df.loc[threshold_df.rmsd <= 5]
            substrate_avg[regime].append(hits.shape[0]/threshold_df.shape[0]*100)
            substrate_counts[regime].append(threshold_df.shape[0])

    all_seq_avg.append(copy(seq_avg))
    all_seq_counts.append(copy(seq_counts))
    all_substrate_avg.append(copy(substrate_avg))
    all_substrate_counts.append(copy(substrate_counts))

fig, axes = plt.subplots(
    len(threshold_tools), 2, figsize=(10,10)
)

for i in range(len(threshold_tools)):
    ax1 = axes[i,0]
    ax2 = axes[i,1]

    seq_avg = all_seq_avg[i]
    seq_counts = all_seq_counts[i]
    substrate_avg = all_substrate_avg[i]
    substrate_counts = all_substrate_counts[i]

    res = ax1.grouped_bar(
        seq_avg,
        tick_labels=thresholds,
        group_spacing=1,
    )
    print(seq_avg)
    for container in res.bar_containers:
        name = container.get_label()
        labels = seq_counts.get(name)
        ax1.bar_label(container, labels=labels, padding=3)

    # Add some text for labels, title, etc.
    ax1.set_ylabel('Hitrate (%)')
    if i == len(threshold_tools)-1:
        ax1.set_xlabel("Similarity threshold (%)")
    ax1.set_ylim(0, 100)

    res = ax2.grouped_bar(
        substrate_avg,
        tick_labels=thresholds,
        group_spacing=1,
    )
    for container in res.bar_containers:
        name = container.get_label()
        labels = substrate_counts.get(name)
        ax2.bar_label(container, labels=labels, padding=3)

    # Add some text for labels, title, etc.
    if i == len(threshold_tools)-1:
        ax2.set_xlabel("Similarity threshold (%)")
    ax2.text(-.15, 1.1, threshold_tools[i], fontsize=15, transform=ax2.transAxes)
    # if i == 0:
    ax2.legend(loc='upper right', ncols=2)
    ax1.legend(loc='upper right', ncols=2)
    ax2.set_ylim(0, 100)
plt.tight_layout()
plt.savefig(f"./figures/clustered_DB.jpg", dpi=600, bbox_inches='tight')
plt.show()
plt.close()
```

```python
from repo import get_ec_number, METAL_IONS, get_metal_info, METAL_ELEMENTS_MAP
import tqdm
import json

pdb_paths = glob.glob("../../databases/QBioLiP/nr_lig/*.pdb")
pdb_ids = [p.split("/")[-1].split("_")[0].upper() for p in pdb_paths if "randconf" not in p]
ri_df = pandas.read_csv("../../databases/QBioLiP/relevant_interactions.csv")
cofactors = json.load(open("../../databases/cofactors.json"))
all_cofactors = [ccd for c in cofactors.values() for ccd in c[0]["cofactors"]]
if not os.path.exists("./pdb_ec_number.pkl"):
    pdb_ec_number = {
        pdb_id:get_ec_number(pdb_id) for pdb_id in tqdm.tqdm(pdb_ids)
    }
    pickle.dump(pdb_ec_number, open("./pdb_ec_number.pkl", "wb"))
else:
    pdb_ec_number = pickle.load(open("./pdb_ec_number.pkl", "rb"))
if not os.path.exists("./pdb_metal_info.pkl"):
    pdb_metal_info = {
        pdb_id:get_metal_info(pdb_id,ri_df.loc[
            (ri_df["PDB ID"] == pdb_id.lower()) &
            (ri_df["Ligand ID"].isin(list(METAL_IONS)))
        ]["Ligand ID"].tolist()) 
        for pdb_id in tqdm.tqdm(pdb_ids)
    }
    pickle.dump(pdb_metal_info, open("./pdb_metal_info.pkl", "wb"))
else:
    pdb_metal_info = pickle.load(open("./pdb_metal_info.pkl", "rb"))
print(pdb_metal_info)
```

```python
group_metals = {
    "A": {},
    "B": {},
    "C": {}
}
group_ec_numbers = {
    "A": {},
    "B": {},
    "C": {}
}
group_geometries = {
    "A": {},
    "B": {},
    "C": {}
}
group_ligands = {
    "A": {},
    "B": {},
    "C": {}
}
for pdb_id in pdb_ids:
    group = df.loc[df.pdb_id == pdb_id.lower()].group.item()
    sub_df = ri_df[
        (ri_df["PDB ID"] == pdb_id.lower()) &
        (ri_df["Ligand ID"].isin(list(METAL_IONS) + all_cofactors))
    ]
    pdb_metals = sub_df.loc[
        sub_df["Ligand ID"].isin(METAL_IONS)
    ]["Ligand ID"].values.unique()
    geometries, ligands = pdb_metal_info[pdb_id]
    for m in pdb_metals:
        metal_name = METAL_ELEMENTS_MAP[m[:2]]["symbol"]
        m_count = group_metals[group].get(metal_name)
        m_count = 0 if m_count is None else m_count
        group_metals[group][metal_name] = m_count + 1
        if group == "A":
            print("pdb of group A:")
            print(sub_df["PDB ID"].unique())
    for ec_number in pdb_ec_number[pdb_id]:
        ec_count = group_ec_numbers[group].get(ec_number[0])
        ec_count = 0 if ec_count is None else ec_count
        group_ec_numbers[group][ec_number[0]] = ec_count + 1
    for geometry in geometries:
        g_count = group_geometries[group].get(geometry)
        g_count = 0 if g_count is None else g_count
        group_geometries[group][geometry] = g_count + 1
    if len(ligands)>0:
        interactor = "ligand/residue"
        if all([i == "ligand" for i in ligands]):
            interactor = "ligand"
        if all([i == "residue" for i in ligands]):
            interactor = "residue"
        i_count = group_ligands[group].get(interactor)
        i_count = 0 if i_count is None else i_count
        group_ligands[group][interactor] = i_count + 1

```

```python
from scipy import stats
def compute_significance(ref, test):
    res = stats.ttest_ind(ref, test, equal_var=False)
    p_value = res.pvalue
    significance = ""
    if p_value < 0.001:
        significance = "***"
    elif p_value < 0.01:
        significance =  "**"
    elif p_value < 0.05:
        significance =  "*"
    return significance
```
